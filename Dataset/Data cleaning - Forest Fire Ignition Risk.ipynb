```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Algerian_forest_fires_dataset_UPDATE.csv", header=1)
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.columns = df.columns.str.strip()
df.columns
```

```python
df['Classes'].unique()
```

```python
df['Classes'] = df['Classes'].str.strip()
df['Classes'].unique()
```

```python
df.isnull().sum()
```

```python
df = df[df['Classes'] != 'Classes']
```

```python
df['Classes'].unique()
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
df['Classes'] = df['Classes'].map({'not fire':0, 'fire':1})
```

```python
df['Classes'].unique()
```

```python
for column in df.columns:
    if column != 'Classes':
        df[column] = pd.to_numeric(df[column])
```

```python
df.info()
df.shape
```

```python
df.to_csv("cleaned_forest_fire_dataset.csv", index=False)
```
